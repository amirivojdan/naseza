```python
import json

file_path = 'naseza.json'
with open(file_path, 'r', encoding='utf-8') as file:
    data = json.load(file)
```

```python
from collections import Counter

# Assuming each 'label' is a list, and contains either 'Offensive' or 'Normal'
label_counts = Counter()
for doc in data:
    for label in doc['label']:
        label_counts[label] += 1

print("Offensive samples:", label_counts.get('Offensive', 0))
print("Normal samples:", label_counts.get('Normal', 0))
print("Total samples:", sum(label_counts.values()))
```
